# 2. Quantum Coin Tossing

## Objective

The objective of quantum coin tossing is to generate a random binary outcome using the probabilistic nature of quantum measurement.

A single qubit is prepared in an equal superposition of the states `|0⟩` and `|1⟩`. When the qubit is measured, either `0` or `1` is obtained.

## Theory

Initially, the qubit is in the state:

\[
|0\rangle
\]

A Hadamard gate is applied to create an equal superposition:

\[
H|0\rangle =
\frac{|0\rangle + |1\rangle}{\sqrt{2}}
\]

Therefore, the probability of measuring `0` is:

\[
P(0)=\frac{1}{2}
\]

and the probability of measuring `1` is:

\[
P(1)=\frac{1}{2}
\]

Thus, the qubit behaves like a quantum coin.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
```

```python
coin = QuantumCircuit(1, 1)

coin.h(0)
coin.measure(0, 0)

print(coin)
```

```python
simulator = AerSimulator()

job = simulator.run(coin, shots=1000)
result = job.result()

counts = result.get_counts()

print("Measurement results:")
print(counts)
```

```python
total = sum(counts.values())

print("Probabilities:")

for bit, count in sorted(counts.items()):
    probability = count / total * 100
    print(f"{bit}: {probability:.2f}%")
```

## Experimental Results

The quantum coin-tossing circuit was executed using 1000 shots.

### Run 1

- `0` = 512 measurements → **51.20%**
- `1` = 488 measurements → **48.80%**

### Run 2

- `0` = 491 measurements → **49.10%**
- `1` = 509 measurements → **50.90%**

The results are close to the theoretical probability of 50% for each outcome. Small variations occur because quantum measurement is probabilistic and only a finite number of shots are used.

## Conclusion

Quantum coin tossing was successfully implemented using Qiskit.

The Hadamard gate created an equal superposition, and measurement produced approximately equal probabilities for the outcomes `0` and `1`.

This experiment demonstrates how quantum superposition can be used to generate random binary outcomes.

# 3. Quantum Secret Sharing

## Objective

Quantum Secret Sharing (QSS) is a quantum cryptographic technique in which a secret is distributed among multiple participants.

In this notebook, we demonstrate the quantum-correlation component of a three-party QSS scheme using a three-qubit GHZ state.

The experiment shows that the three qubits produce strongly correlated measurement results.

## Theory

The experiment begins with three qubits in the state:

\[
|000\rangle
\]

A Hadamard gate is applied to the first qubit:

\[
H|0\rangle =
\frac{|0\rangle + |1\rangle}{\sqrt{2}}
\]

The first qubit is then connected to the second and third qubits using CNOT gates.

This produces the three-qubit GHZ state:

\[
|\mathrm{GHZ}\rangle =
\frac{|000\rangle + |111\rangle}{\sqrt{2}}
\]

When the qubits are measured, the expected outcomes are `000` and `111`, with approximately equal probability.

The important feature is the strong correlation between the three qubits.

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
```

```python
secret = QuantumCircuit(3, 3)

# Create a three-qubit GHZ state
secret.h(0)
secret.cx(0, 1)
secret.cx(0, 2)

print(secret)
```

```python
secret.measure([0, 1, 2], [0, 1, 2])

print(secret)
```

```python
simulator = AerSimulator()

job = simulator.run(secret, shots=1000)
result = job.result()

counts = result.get_counts()

print("Measurement results:")
print(counts)
```

```python
total = sum(counts.values())

print("Probabilities:")

for state, count in sorted(counts.items()):
    probability = count / total * 100
    print(f"{state}: {probability:.2f}%")
```

## Experimental Results

The three-qubit GHZ circuit was executed using 1000 shots.

### Run 1

- `000` = 512 measurements → **51.20%**
- `111` = 488 measurements → **48.80%**

### Run 2

- `000` = 467 measurements → **46.70%**
- `111` = 533 measurements → **53.30%**

The results are concentrated on `000` and `111`, demonstrating strong correlation between the three qubits.

The small difference from exactly 50% is due to the probabilistic nature of quantum measurement and the finite number of shots.

## Interpretation

The GHZ state creates strong correlations among the three qubits.

The three qubits do not produce all eight possible three-bit outcomes with equal probability. Instead, the measurement is concentrated on:

- `000`
- `111`

If the first qubit is measured as `0`, the other two are also observed as `0`.

If the first qubit is measured as `1`, the other two are also observed as `1`.

This correlation is an important quantum resource for quantum secret-sharing protocols.

## Conclusion

A three-qubit GHZ state was successfully implemented and simulated using Qiskit.

The experimental results showed approximately equal probabilities for the correlated states `000` and `111`.

This experiment demonstrates the quantum-correlation component used in quantum secret-sharing schemes.

Note: This notebook demonstrates the GHZ-state foundation of QSS. A complete QSS protocol would additionally define secret encoding, distribution, reconstruction, and security against an adversary.
