# 1. Fundamental Quantum Concepts

Before implementing quantum communication protocols, it is
important to understand the basic concepts used by these
protocols.

## 1.1 Qubit

A classical bit can have a value of either 0 or 1.

A quantum bit, called a qubit, can exist in a superposition
of the states |0⟩ and |1⟩.

The general state of a qubit is:

|ψ⟩ = α|0⟩ + β|1⟩

where α and β are probability amplitudes.

The probabilities must satisfy:

|α|² + |β|² = 1.

## 1.2 Superposition

Superposition means that a qubit can exist as a combination
of multiple quantum states before measurement.

The Hadamard gate can transform |0⟩ into:

|+⟩ = (|0⟩ + |1⟩) / √2

When this state is measured in the computational basis,
the results 0 and 1 occur with approximately equal
probability.

## 1.3 Measurement

Measurement converts quantum information into a classical
result.

For example, measuring:

|+⟩ = (|0⟩ + |1⟩) / √2

in the computational basis produces either 0 or 1.

## 1.4 Entanglement

Entanglement is a strong quantum correlation between
quantum systems.

One important Bell state is:

|Φ⁺⟩ = (|00⟩ + |11⟩) / √2

Entanglement is an important resource in quantum
teleportation and entanglement-based quantum key
distribution.

## 1.5 Quantum Bit Error Rate (QBER)

QBER is used to measure the error rate in a quantum
communication protocol.

QBER is calculated as:

QBER = Number of errors / Number of compared bits

A higher QBER can indicate noise, implementation errors,
or possible eavesdropping.

# 3. BB84 Quantum Key Distribution

## Introduction

BB84 is a quantum key distribution (QKD) protocol proposed by
Charles Bennett and Gilles Brassard in 1984.

The main purpose of BB84 is to allow two parties, Alice and Bob,
to establish a shared secret key using quantum states.

The security of BB84 is based on fundamental properties of
quantum mechanics.

In particular:

- Quantum states cannot be copied perfectly.
- Measuring a quantum state can disturb it.
- Non-orthogonal quantum states cannot always be distinguished
  perfectly.

## Participants

The protocol involves:

- Alice — sender
- Bob — receiver
- Eve — potential eavesdropper

Alice sends quantum states to Bob. Bob measures those states
using randomly selected measurement bases.

After the quantum transmission, Alice and Bob publicly compare
their measurement bases, but they do not reveal their measured
bit values.

They keep only the measurements where their bases agree.
This process is called sifting.

The remaining bits form the raw key.

## BB84 Bases

BB84 commonly uses two bases:

### Computational basis

The computational basis contains:

| State | Bit |
|-------|-----|
| |0⟩ | 0 |
| |1⟩ | 1 |

### Hadamard basis

The Hadamard basis contains:

| State | Bit |
|-------|-----|
| |+⟩ | 0 |
| |−⟩ | 1 |

The two bases are incompatible, which is important for detecting
eavesdropping.

## BB84 Protocol — Step-by-Step Process

### Step 1: Alice generates random bits

Alice generates a random sequence of 0s and 1s.

Example:

0 1 1 0 1 0 0 1

### Step 2: Alice chooses random bases

For every bit, Alice randomly chooses either:

- Computational basis
- Hadamard basis

### Step 3: Alice encodes the bits

Alice prepares quantum states according to the selected bit
and basis.

For example:

Computational basis:

0 → |0⟩
1 → |1⟩

Hadamard basis:

0 → |+⟩
1 → |−⟩

### Step 4: Bob chooses random bases

Bob independently chooses a random basis for each received
quantum state.

### Step 5: Bob measures

Bob measures each quantum state using his selected basis.

If Bob chooses the same basis as Alice, he obtains the correct
bit in the ideal case.

If Bob chooses a different basis, the result is not useful for
forming the raw key.

### Step 6: Basis reconciliation

Alice and Bob publicly compare their basis choices.

They keep only positions where their bases are equal.

### Step 7: Sifting

The remaining bits form the raw key.

### Step 8: Error analysis

Alice and Bob can compare a subset of their key to estimate
the Quantum Bit Error Rate (QBER).

A significant QBER can indicate noise or possible
eavesdropping.

```python
import random

# Number of quantum bits to generate
N = 20

# Alice generates random bits
alice_bits = [random.randint(0, 1) for _ in range(N)]

# Alice randomly chooses a basis for each bit
# 0 = computational basis
# 1 = Hadamard basis
alice_bases = [random.randint(0, 1) for _ in range(N)]

print("Alice bits:")
print(alice_bits)

print("\nAlice bases:")
print(alice_bases)
```

## Alice's BB84 Quantum Encoding

Alice encodes each classical bit into a quantum state using
one of two measurement bases.

### Computational basis

If Alice chooses basis 0:

- Bit 0 → |0⟩
- Bit 1 → |1⟩

### Hadamard basis

If Alice chooses basis 1:

- Bit 0 → |+⟩
- Bit 1 → |−⟩

The Hadamard basis states are:

|+⟩ = (|0⟩ + |1⟩) / √2

|−⟩ = (|0⟩ − |1⟩) / √2

Therefore, Alice's bit and basis determine which quantum state
is transmitted to Bob.

```python
from qiskit import QuantumCircuit

def encode_bb84(bit, basis):
    """
    Encode one classical bit using the BB84 protocol.

    basis = 0 → computational basis
    basis = 1 → Hadamard basis
    """

    circuit = QuantumCircuit(1, 1)

    # Encode bit value
    if bit == 1:
        circuit.x(0)

    # Encode measurement basis
    if basis == 1:
        circuit.h(0)

    return circuit
```

```python
print("BB84 encoding examples")

for bit in [0, 1]:
    for basis in [0, 1]:

        circuit = encode_bb84(bit, basis)

        print(
            f"\nBit = {bit}, Basis = {basis}"
        )
        print(circuit)
```

## Bob's Random Measurement Bases

Bob does not know which basis Alice used for each qubit.

Therefore, Bob independently chooses a random basis for every
received qubit.

The basis choices are:

- 0 → Computational basis
- 1 → Hadamard basis

Bob then measures each received quantum state using his selected
basis.

Bob's basis choices are independent of Alice's choices.

```python
# Bob independently chooses random bases
bob_bases = [random.randint(0, 1) for _ in range(N)]

print("Alice bases:")
print(alice_bases)

print("\nBob bases:")
print(bob_bases)
```

## Bob's Measurement Basis

Bob must choose a measurement basis before measuring each
qubit.

### If Bob chooses the computational basis

Bob measures directly in:

{|0⟩, |1⟩}

The measurement result is interpreted as:

0 → bit 0

1 → bit 1

### If Bob chooses the Hadamard basis

Bob first applies a Hadamard gate and then measures in the
computational basis.

This converts:

|+⟩ → |0⟩

|−⟩ → |1⟩

Therefore, the Hadamard gate allows Bob to measure in the
Hadamard basis.

```python
from qiskit_aer import AerSimulator

simulator = AerSimulator()


def measure_bb84(bit, alice_basis, bob_basis):
    """
    Simulate transmission of one BB84 quantum state.

    Returns Bob's measurement result.
    """

    # Alice encodes the state
    circuit = encode_bb84(bit, alice_basis)

    # Bob chooses Hadamard basis
    if bob_basis == 1:
        circuit.h(0)

    # Measure the qubit
    circuit.measure(0, 0)

    # Run one shot
    result = simulator.run(circuit, shots=1).result()

    counts = result.get_counts()

    return int(next(iter(counts)))
```

```python
# Test transmission

alice_bit = 1
alice_basis = 0
bob_basis = 0

bob_result = measure_bb84(
    alice_bit,
    alice_basis,
    bob_basis
)

print("Alice bit:", alice_bit)
print("Alice basis:", alice_basis)
print("Bob basis:", bob_basis)
print("Bob measurement:", bob_result)
```

```python
bob_results = []

for i in range(N):

    result = measure_bb84(
        alice_bits[i],
        alice_bases[i],
        bob_bases[i]
    )

    bob_results.append(result)

print("Alice bits:")
print(alice_bits)

print("\nAlice bases:")
print(alice_bases)

print("\nBob bases:")
print(bob_bases)

print("\nBob measurements:")
print(bob_results)
```

## BB84 Basis Reconciliation and Sifting

After quantum transmission, Alice and Bob publicly compare
their basis choices.

They do not reveal their measured bit values during this step.

For every transmitted qubit:

- If Alice's basis equals Bob's basis, the result is kept.
- If Alice's basis differs from Bob's basis, the result is discarded.

The remaining bits form the raw key.

This process is called sifting.

In an ideal BB84 implementation, the retained bits should
match between Alice and Bob.

```python
# Sifting process

alice_raw_key = []
bob_raw_key = []
kept_positions = []

for i in range(N):

    # Keep only positions where Alice and Bob
    # used the same measurement basis
    if alice_bases[i] == bob_bases[i]:

        alice_raw_key.append(alice_bits[i])
        bob_raw_key.append(bob_results[i])
        kept_positions.append(i)

print("Kept positions:")
print(kept_positions)

print("\nAlice raw key:")
print(alice_raw_key)

print("\nBob raw key:")
print(bob_raw_key)

print("\nRaw key length:")
print(len(alice_raw_key))
```

```python
# Compare Alice's and Bob's raw keys

errors = sum(
    a != b
    for a, b in zip(alice_raw_key, bob_raw_key)
)

print("Alice raw key:")
print(alice_raw_key)

print("\nBob raw key:")
print(bob_raw_key)

print("\nErrors:")
print(errors)

if errors == 0:
    print("\nSUCCESS!")
    print("Alice and Bob have identical raw keys.")
else:
    print("\nERRORS DETECTED!")
```

## Quantum Bit Error Rate (QBER)

The Quantum Bit Error Rate (QBER) measures the fraction of
bits that differ between Alice's and Bob's raw keys.

It is calculated as:

QBER = Number of differing bits / Total number of compared bits

A low QBER indicates that the quantum channel is relatively
clean.

A high QBER can indicate:

- Eavesdropping
- Noise
- Measurement errors
- Imperfections in the physical quantum system

For an ideal simulator without eavesdropping or noise, the
expected QBER is approximately zero.

```python
# Calculate QBER

if len(alice_raw_key) > 0:
    qber = errors / len(alice_raw_key)
else:
    qber = 0

print("Raw key length:", len(alice_raw_key))
print("Errors:", errors)
print("QBER:", qber)
print("QBER percentage:", qber * 100, "%")
```

# 4. BB84 with Eavesdropping

## Introduction

The security of BB84 can be demonstrated by introducing a
potential eavesdropper, Eve, into the communication channel.

Eve attempts to intercept the quantum states sent by Alice
and then forwards states to Bob.

Because Eve does not know which measurement basis Alice used,
she must choose a basis independently.

If Eve chooses the wrong basis, her measurement can disturb
the quantum state.

When Bob later measures the disturbed state, errors may appear
in the raw key.

Alice and Bob can estimate these errors using the Quantum Bit
Error Rate (QBER).

A significantly increased QBER can indicate the presence of
eavesdropping.

## Intercept-and-Resend Attack

The attack consists of three main steps:

1. Alice prepares and sends a BB84 quantum state.
2. Eve intercepts the state, chooses a random basis, and
   measures it.
3. Eve prepares a new quantum state corresponding to her
   measurement result and sends it to Bob.

Eve does not have access to Alice's original basis during the
quantum transmission.

Therefore, Eve sometimes chooses the incorrect basis.

This disturbance can be detected by Alice and Bob through
QBER analysis.

```python
# Eve independently chooses a random basis for every qubit

eve_bases = [random.randint(0, 1) for _ in range(N)]

print("Alice bases:")
print(alice_bases)

print("\nEve bases:")
print(eve_bases)

print("\nBob bases:")
print(bob_bases)
```

```python
def eve_intercept_and_resend(bit, alice_basis, eve_basis):
    """
    Eve intercepts Alice's BB84 state.

    Eve chooses her own measurement basis, measures the state,
    and obtains a classical bit.

    Eve then resends a new BB84 state based on her result.
    """

    # Eve measures Alice's transmitted state
    eve_result = measure_bb84(
        bit,
        alice_basis,
        eve_basis
    )

    return eve_result
```

```python
bob_results_eve = []
eve_results = []

for i in range(N):

    # Eve intercepts Alice's state
    eve_result = eve_intercept_and_resend(
        alice_bits[i],
        alice_bases[i],
        eve_bases[i]
    )

    eve_results.append(eve_result)

    # Eve resends her measured bit using Eve's basis
    bob_result = measure_bb84(
        eve_result,
        eve_bases[i],
        bob_bases[i]
    )

    bob_results_eve.append(bob_result)


print("Alice bits:")
print(alice_bits)

print("\nAlice bases:")
print(alice_bases)

print("\nEve bases:")
print(eve_bases)

print("\nEve measurements:")
print(eve_results)

print("\nBob bases:")
print(bob_bases)

print("\nBob measurements:")
print(bob_results_eve)
```

```python
alice_raw_key_eve = []
bob_raw_key_eve = []
kept_positions_eve = []

for i in range(N):

    # Alice and Bob publicly compare their bases
    if alice_bases[i] == bob_bases[i]:

        alice_raw_key_eve.append(alice_bits[i])
        bob_raw_key_eve.append(bob_results_eve[i])
        kept_positions_eve.append(i)


print("Kept positions:")
print(kept_positions_eve)

print("\nAlice raw key:")
print(alice_raw_key_eve)

print("\nBob raw key:")
print(bob_raw_key_eve)
```

```python
errors_eve = sum(
    a != b
    for a, b in zip(alice_raw_key_eve, bob_raw_key_eve)
)

if len(alice_raw_key_eve) > 0:
    qber_eve = errors_eve / len(alice_raw_key_eve)
else:
    qber_eve = 0

print("Raw key length:")
print(len(alice_raw_key_eve))

print("\nErrors:")
print(errors_eve)

print("\nQBER:")
print(qber_eve)

print("\nQBER percentage:")
print(qber_eve * 100, "%")
```

## BB84 Eavesdropping — Interpretation

In the intercept-and-resend attack, Eve does not know Alice's
measurement basis.

When Eve chooses the same basis as Alice, she can correctly
identify the transmitted bit without introducing a basis
measurement error.

When Eve chooses a different basis, her measurement produces a
random result relative to Alice's original state.

Eve then resends a state based on her measurement.

If Alice and Bob later use the same basis, the disturbance caused
by Eve can result in a mismatch between their raw-key bits.

Therefore, the QBER becomes greater than zero.

A significant QBER provides evidence that the communication
channel may have been disturbed.

This demonstrates the fundamental security principle of BB84:
attempting to obtain information about unknown quantum states can
disturb those states and reveal the presence of an eavesdropper.

# 5. E91 Quantum Key Distribution

## Introduction

E91 is a quantum key distribution protocol proposed by
Artur Ekert in 1991.

Unlike BB84, which prepares and sends individual quantum
states, E91 uses entangled quantum pairs.

An entangled pair is shared between Alice and Bob.

Alice and Bob independently choose measurement settings and
measure their respective qubits.

For certain combinations of measurement settings, their
measurement results exhibit strong quantum correlations.

These correlations can be used to generate a shared secret key.

The protocol can also use the violation of a Bell inequality
as evidence that the observed correlations have a quantum
origin.

## Main Components

The E91 protocol consists of:

1. Entangled pair generation
2. Distribution of the entangled qubits
3. Random measurement settings
4. Measurement of the qubits
5. Correlation analysis
6. Key generation
7. Bell-inequality verification

## Security Principle

The security of E91 is related to the non-classical correlations
produced by entanglement.

An eavesdropper attempting to gain information about the
quantum states can disturb these correlations.

Therefore, Alice and Bob can analyze their measurement
correlations to detect possible interference.

## E91 Entanglement

A commonly used entangled state is the Bell state:

|Φ⁺⟩ = (|00⟩ + |11⟩) / √2

This means that when Alice and Bob measure the two qubits in
the same computational basis, their results are correlated.

For example:

Alice = 0 → Bob = 0

Alice = 1 → Bob = 1

The individual measurement result is random, but the relationship
between the two results is correlated.

```python
from qiskit import QuantumCircuit

e91_bell = QuantumCircuit(2, 2)

# Create superposition
e91_bell.h(0)

# Create entanglement
e91_bell.cx(0, 1)

print(e91_bell)
```

```python
e91_measurement = QuantumCircuit(2, 2)

# Create Bell state
e91_measurement.h(0)
e91_measurement.cx(0, 1)

# Measure both qubits
e91_measurement.measure(0, 0)
e91_measurement.measure(1, 1)

print(e91_measurement)
```

```python
from qiskit_aer import AerSimulator

simulator = AerSimulator()

job = simulator.run(
    e91_measurement,
    shots=1024
)

result = job.result()
counts = result.get_counts()

print("Bell-state measurement results:")
print(counts)
```

## Bell-State Experimental Result

The Bell-state experiment produced approximately equal numbers
of |00⟩ and |11⟩ outcomes.

The |01⟩ and |10⟩ outcomes are ideally absent.

This demonstrates the correlation produced by the entangled Bell
state.

The experiment illustrates an important property of quantum
entanglement:

- Individual measurement results are probabilistic.
- Joint measurement results are strongly correlated.

These correlations form the foundation of the E91 quantum key
distribution protocol.

In the complete E91 protocol, Alice and Bob use multiple
measurement settings. Some settings are used for key generation,
while others are used to test Bell-inequality correlations.

## E91 Measurement Settings

In the E91 protocol, Alice and Bob randomly select measurement
settings for their entangled particles.

Different measurement settings produce different correlations.

Some measurement combinations are used for generating the
shared key.

Other combinations are used to test the strength of the
quantum correlations.

The Bell-inequality test is important because classical
correlations cannot reproduce the same correlation strength
under the relevant assumptions.

Therefore, observing an appropriate Bell-inequality violation
provides evidence of non-classical quantum correlations.

## CHSH Bell Test

The CHSH form of the Bell inequality uses two measurement
settings for Alice and two measurement settings for Bob.

The CHSH parameter is calculated as:

S = E(a,b) + E(a,b') + E(a',b) - E(a',b')

where E represents the correlation between Alice's and Bob's
measurement results.

For local classical theories:

|S| ≤ 2

Quantum mechanics can produce values up to:

|S| = 2√2

A value greater than 2 demonstrates a violation of the
classical CHSH bound.

```python
import math
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

def create_bell_pair():
    circuit = QuantumCircuit(2, 2)

    # Create Bell state
    circuit.h(0)
    circuit.cx(0, 1)

    return circuit
```

```python
def measure_in_basis(circuit, qubit, angle):
    """
    Rotate a qubit before computational-basis measurement.

    The angle controls the measurement basis.
    """

    circuit.ry(-2 * angle, qubit)
```

```python
def calculate_correlation(angle_a, angle_b, shots=2048):

    circuit = create_bell_pair()

    # Alice's measurement rotation
    measure_in_basis(circuit, 0, angle_a)

    # Bob's measurement rotation
    measure_in_basis(circuit, 1, angle_b)

    # Measure both qubits
    circuit.measure(0, 0)
    circuit.measure(1, 1)

    result = simulator.run(
        circuit,
        shots=shots
    ).result()

    counts = result.get_counts()

    # Convert measurement results into ±1 values
    correlation = 0

    for state, count in counts.items():

        # Qiskit displays classical bits as c1 c0
        alice_bit = int(state[-1])
        bob_bit = int(state[-2])

        alice_value = 1 if alice_bit == 0 else -1
        bob_value = 1 if bob_bit == 0 else -1

        correlation += (
            alice_value *
            bob_value *
            count
        )

    return correlation / shots
```

```python
# Measurement settings

a = 0
a_prime = math.pi / 4

b = math.pi / 8
b_prime = -math.pi / 8

print("Alice settings:")
print("a =", a)
print("a' =", a_prime)

print("\nBob settings:")
print("b =", b)
print("b' =", b_prime)
```

```python
E_ab = calculate_correlation(a, b)
E_ab_prime = calculate_correlation(a, b_prime)
E_a_prime_b = calculate_correlation(a_prime, b)
E_a_prime_b_prime = calculate_correlation(
    a_prime,
    b_prime
)

print("E(a,b)       =", E_ab)
print("E(a,b')      =", E_ab_prime)
print("E(a',b)      =", E_a_prime_b)
print("E(a',b')     =", E_a_prime_b_prime)
```

```python
S = (
    E_ab
    + E_ab_prime
    + E_a_prime_b
    - E_a_prime_b_prime
)

print("CHSH S =", S)
print("|S| =", abs(S))

if abs(S) > 2:
    print("\nBell inequality violated.")
    print("This indicates non-classical quantum correlations.")
else:
    print("\nNo Bell inequality violation detected.")
```

# 6. B92 Quantum Key Distribution

## Introduction

B92 is a quantum key distribution protocol proposed by Charles
Bennett in 1992.

Unlike BB84, which uses four quantum states from two bases,
B92 uses only two non-orthogonal quantum states.

The two states cannot be perfectly distinguished from one
another.

This property is important for quantum security because an
attempt to determine the transmitted state can introduce
uncertainty or errors.

In this project, B92 is implemented using the Perceval
photonic quantum computing framework.

## Alice's Encoding

Alice encodes classical bits using two non-orthogonal states:

Bit 0 → |0⟩

Bit 1 → |+⟩

where:

|+⟩ = (|0⟩ + |1⟩) / √2

Because |0⟩ and |+⟩ are non-orthogonal, Bob cannot always
distinguish them perfectly.

Therefore, some measurement results are inconclusive.

## Bob's Measurement

Bob performs measurements designed to identify the transmitted
state unambiguously.

Measurement A:

Detect |1⟩

→ conclusive result for Alice's bit 1

Measurement B:

Detect |−⟩

→ conclusive result for Alice's bit 0

Other detector outcomes are inconclusive and are discarded.

Only conclusive rounds are used to construct the raw key.

```python
import perceval as pcvl

print("Perceval version:", pcvl.__version__)
```

## Alice's B92 Quantum States

In the Perceval path-encoding representation:

| Logical state | Perceval representation |
|---|---|
| |0⟩ | |1,0⟩ |
| |1⟩ | |0,1⟩ |

Alice's B92 encoding is:

Bit 0 → |0⟩

Bit 1 → |+⟩

The |+⟩ state is produced by applying a 50:50 beam splitter
to the appropriate input state.

The resulting state is a superposition of the two optical
modes.

```python
# Alice's logical |0> state
state_zero = pcvl.BasicState([1, 0])

print("Alice bit 0:")
print("|0> =", state_zero)
```

```python
# Prepare |+> using a 50:50 beam splitter

plus_circuit = pcvl.Circuit(2)
plus_circuit.add(0, pcvl.BS())

plus_processor = pcvl.Processor(
    "SLOS",
    plus_circuit
)

plus_processor.with_input(state_zero)

plus_sampler = pcvl.algorithm.Sampler(
    plus_processor
)

plus_result = plus_sampler.probs()

print("Alice bit 1 -> |+>:")
print(plus_result)
```

## Interpretation of the |+⟩ Preparation

The Perceval simulation produces two possible output states,
each with probability 0.5.

This demonstrates that the beam splitter creates a quantum
superposition of the two optical modes.

Therefore, the experiment verifies the preparation of the
superposition state used for Alice's B92 bit-1 encoding.

The result is probabilistic because quantum measurement does
not determine a single output before measurement.

## B92 Transmission

In a complete B92 transmission:

1. Alice generates a random classical bit.
2. Alice encodes the bit as |0⟩ or |+⟩.
3. Bob randomly selects one of the B92 measurement procedures.
4. Bob obtains either a conclusive or inconclusive result.
5. Alice and Bob keep only conclusive rounds.
6. The retained bits form the raw key.

The following simulation models this process for multiple
transmission rounds.

```python
import random

N_B92 = 20

alice_bits_b92 = [
    random.randint(0, 1)
    for _ in range(N_B92)
]

alice_raw_key_b92 = []
bob_raw_key_b92 = []

print("=" * 70)
print("B92 QUANTUM KEY DISTRIBUTION")
print("=" * 70)

print("\nRound-by-round results")
print("-" * 70)

for i in range(N_B92):

    alice_bit = alice_bits_b92[i]

    # Bob randomly chooses measurement A or B
    bob_measurement = random.randint(0, 1)

    bob_result = None

    # Measurement A:
    # Detect |1> -> conclusive bit 1
    if bob_measurement == 0:

        if alice_bit == 1:
            # Simplified probabilistic conclusive detection
            if random.random() < 0.5:
                bob_result = 1

    # Measurement B:
    # Detect |-> -> conclusive bit 0
    else:

        if alice_bit == 0:
            # Simplified probabilistic conclusive detection
            if random.random() < 0.5:
                bob_result = 0

    print(
        f"Round {i+1:02d}: "
        f"Alice={alice_bit} "
        f"Bob measurement={bob_measurement} "
        f"Bob result={bob_result}"
    )

    # Keep only conclusive results
    if bob_result is not None:

        alice_raw_key_b92.append(alice_bit)
        bob_raw_key_b92.append(bob_result)
```

```python
print("\n" + "=" * 70)
print("B92 SIFTING RESULT")
print("=" * 70)

print("\nAlice raw key:")
print(alice_raw_key_b92)

print("\nBob raw key:")
print(bob_raw_key_b92)

print("\nKey length:")
print(len(alice_raw_key_b92))
```

```python
errors_b92 = sum(
    a != b
    for a, b in zip(
        alice_raw_key_b92,
        bob_raw_key_b92
    )
)

print("Errors:", errors_b92)

if errors_b92 == 0:
    print("\nSUCCESS!")
    print("Alice and Bob have the same raw key.")
else:
    print("\nERRORS DETECTED!")
```

# 7. B92 with Eavesdropping

## Intercept-and-Resend Attack

To study the security of B92, an eavesdropper named Eve is
introduced into the communication channel.

Eve intercepts Alice's quantum state and attempts to determine
which of the two non-orthogonal states Alice transmitted.

Because the B92 states are non-orthogonal, Eve cannot always
identify the state without uncertainty.

Eve then prepares and sends a new state toward Bob based on
her measurement result.

This process can disturb the quantum communication.

Alice and Bob can detect this disturbance by comparing a subset
of their retained key bits and calculating the Quantum Bit
Error Rate (QBER).

A non-zero or significantly increased QBER can indicate that
the communication channel has been disturbed.

```python
# Number of B92 transmission rounds

N_B92_EVE = 50

alice_bits_b92_eve = [
    random.randint(0, 1)
    for _ in range(N_B92_EVE)
]

alice_raw_key_eve_b92 = []
bob_raw_key_eve_b92 = []

eve_measurements_b92 = []

print("=" * 70)
print("B92 QUANTUM KEY DISTRIBUTION - EAVESDROPPING")
print("=" * 70)

print("\nRound-by-round results")
print("-" * 70)

for i in range(N_B92_EVE):

    alice_bit = alice_bits_b92_eve[i]

    # Eve chooses one of the two measurement procedures
    eve_measurement = random.randint(0, 1)

    eve_result = None

    # Eve measurement A:
    # attempts to detect |1>
    if eve_measurement == 0:

        if alice_bit == 1:
            if random.random() < 0.5:
                eve_result = 1

    # Eve measurement B:
    # attempts to detect |->
    else:

        if alice_bit == 0:
            if random.random() < 0.5:
                eve_result = 0

    # Only conclusive Eve measurements are used
    if eve_result is not None:

        # Eve resends the state corresponding
        # to her measurement result.
        resent_bit = eve_result

        # Bob performs a random measurement
        bob_measurement = random.randint(0, 1)

        bob_result = None

        if bob_measurement == 0:

            if resent_bit == 1:
                if random.random() < 0.5:
                    bob_result = 1

        else:

            if resent_bit == 0:
                if random.random() < 0.5:
                    bob_result = 0

        # Keep Bob's conclusive result
        if bob_result is not None:

            alice_raw_key_eve_b92.append(alice_bit)
            bob_raw_key_eve_b92.append(bob_result)

            eve_measurements_b92.append(eve_result)

            print(
                f"Round {i+1:02d}: "
                f"Alice={alice_bit} "
                f"Eve={eve_result} "
                f"Bob={bob_result}"
            )
```

```python
print("\n" + "=" * 70)
print("B92 EAVESDROPPING RESULT")
print("=" * 70)

print("\nAlice raw key:")
print(alice_raw_key_eve_b92)

print("\nBob raw key:")
print(bob_raw_key_eve_b92)

raw_key_length_b92 = len(
    alice_raw_key_eve_b92
)

errors_b92_eve = sum(
    a != b
    for a, b in zip(
        alice_raw_key_eve_b92,
        bob_raw_key_eve_b92
    )
)

if raw_key_length_b92 > 0:

    qber_b92_eve = (
        errors_b92_eve /
        raw_key_length_b92
    )

else:

    qber_b92_eve = 0

print("\nRaw key length:")
print(raw_key_length_b92)

print("\nErrors:")
print(errors_b92_eve)

print("\nQBER:")
print(qber_b92_eve)

print("\nQBER percentage:")
print(qber_b92_eve * 100, "%")
```

## B92 Eavesdropping — Interpretation

The intercept-and-resend attack can introduce differences
between Alice's and Bob's retained key bits.

The resulting QBER measures the fraction of mismatched bits:

QBER = Number of errors / Total number of retained bits

A non-zero QBER indicates that the communication channel has
experienced disturbances.

In an actual B92 implementation, the observed QBER depends on
the optical system, measurement apparatus, channel noise,
detector behavior, and the eavesdropping strategy.

Therefore, the simulation demonstrates the security principle
rather than representing a complete physical implementation.

## Key Differences

### 1. Quantum Teleportation

Quantum teleportation is different from the other three protocols because its primary objective is **quantum-state transfer**, not secret-key generation. It uses a shared entangled pair, a Bell-state measurement, and classical communication to reconstruct the quantum state at Bob's side.

### 2. BB84

BB84 is a **prepare-and-measure QKD protocol**. Alice prepares quantum states using two different bases, while Bob randomly chooses a measurement basis. After the quantum transmission, Alice and Bob publicly compare their basis choices and discard incompatible measurements.

### 3. E91

E91 is an **entanglement-based QKD protocol**. Alice and Bob share entangled particles and perform measurements using different settings. Bell-inequality tests, commonly expressed using the CHSH parameter, provide a way to examine whether the observed correlations are consistent with quantum behavior.

### 4. B92

B92 simplifies the state preparation compared with BB84 by using **two non-orthogonal quantum states instead of four states**. Because the states are non-orthogonal, Bob cannot always identify the transmitted state. Therefore, some measurement outcomes are inconclusive and are discarded.

## Overall Comparison

The protocols can therefore be classified as:

**Quantum-state transfer**

→ Quantum Teleportation

**Prepare-and-measure QKD**

→ BB84
→ B92

**Entanglement-based QKD**

→ E91

The experiments in this project demonstrate the different security and communication principles used by these protocols. BB84 and B92 demonstrate key generation and eavesdropping detection through measurement errors, while E91 demonstrates the role of entanglement and Bell correlations. Quantum teleportation demonstrates how an unknown quantum state can be transferred using shared entanglement and classical communication.

## Conclusion of the Comparison

There is no single protocol that is best for every quantum communication application. The appropriate protocol depends on the objective, available quantum hardware, channel characteristics, implementation complexity, and required security properties.

For this project, studying all four protocols provides a progression from basic quantum-state manipulation to practical QKD concepts:

**Quantum state → Entanglement → Measurement → Key generation → Eavesdropping detection**

This comparison also demonstrates why quantum communication is not a single protocol but a collection of techniques based on different quantum-mechanical principles.
